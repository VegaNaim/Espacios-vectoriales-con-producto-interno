### Laboratorio 06: Base y dimensión de un espacio vectorial.
Ya hemos descrito los aspectos generales de un espacio vectorial: es un conjunto de objetos para el cual se definen dos operaciones, a saber: suma y producto por escalar, las cuales cumplen con diez propiedades: cerradura, conmutatividad, asociatividad, distributibilidad, existencia de neutro e inverso para la suma (vectores $\mathbf{0}$ y $-\mathbf{v}$ para cada vector $\mathbf{v}$), además de que el escalar 1 es el idéntico multiplicativo para la operación producto. Además, hemos visto que un espacio vectorial puede generarse por medio de un conjunto finito de elementos de dicho espacio; dicho de otro modo, si cualquier elemento de un espacio vectorial puede representarse por medio de una combinación lineal de un conjunto finito de vectores, entonces se dice que este conjunto genera a todo el espacio vectorial. Escribiremos una definición para este conjunto:

#### Definición. Base.
Un conjunto finito de vectores $$\{\mathbf{v}_{1},\mathbf{v}_{1},\ldots,\mathbf{v}_{n}\}$$ es una base para un espacio vectorial $V$ si
1. $\{\mathbf{v}_{1},\mathbf{v}_{1},\ldots,\mathbf{v}_{n}\}$ genera a $V$.
2. $\{\mathbf{v}_{1},\mathbf{v}_{1},\ldots,\mathbf{v}_{n}\}$ es linealmente independiente


En realidad, un espacio vectorial puede tener más de una base. Cualquier conjunto de $n$ vectores que cumplan estas dos características dadas en la definición anterior califica como base. Ahora, es importante notar que si se tienen dos bases, éstas tienen exactamente la misma cantidad de vectores. Esto nos permitirá escribir la siguiente definición.

#### Definición. Dimensión.
Si el espacio vectorial $V$ tiene una base con un número finito de elementos, entonces la dimensión de $V$ , denotada por $\dim V$, es el número de vectores en todas las bases. Si $V=\{\mathbf{0}\}$ (el espacio nulo), entonces se dice que $V$ tiene dimensión cero.

### Ejercicio 1: la función `LI()`.

Escribe una función que reciba una matriz y determine si sus vectores columna son linealmente independientes.

- Entradas:
  - Un objeto `Matrix` de _SymPy_.
- Salidas:
  - Un valor `bool`
- Especificaciones:
  La función devolverá:
  - Un valor `True` si los vectores columna son L.I.
  - Un valor `False` si los vectores columna son L.D.
  La función deberá tomar la decisión **a partir de la cantidad de pivotes que presente la RREF**.
- Sugerencia: La dimensión de un objeto `Matrix` de _SymPy_ se obtiene a con el método `shape`: el código `matriz.shape` devuelve una tupla con la información `(renglones, columnas)`.
  








```python
# Escribe tu código aquí:

# Creación de vectores de prueba como objeto de sympy
v1 = Matrix([1,2,3,8])
v2 = Matrix([8,6,7,5])
v3 = Matrix([2,5,9,4])

# Creación de la matriz como objeto de sympy
A = Matrix(np.hstack((v1,v2,v3)))

print('Matriz original:')
print(A)
print('Cantidad de renglones y columnas:')
print(A.shape)

# Funcion que determina si el conjunto es linealmente independiente
def LI(A):

  # Variable que almacena el conteo de pivotes
  pivotes = A.rank()

  # Variable que almacena el número de columnas
  columnas = A.shape[1]

  print('Matriz escalonada reducida:')
  print(A.rref())
  print('Cantidad de pivotes:')
  print(pivotes)
  print('Cantidad de columnas:')
  print(columnas)

  # Condición para evaluar si la cantidad de pivotes que hay en la matriz coincide con el numero de columnas
  if pivotes == columnas:
    print('Es un conjunto linealmente independiente')
    return True
  else:
    print('Es un conjunto linealmente dependiente')
    return False

# LLamada a la función anterior
LI(A)
```

#### Parte 2. Decide si es un conjunto generador.
Ahora, se analiza si son un conjunto generador. Para esto, tomaremos un vector arbitrario $\boldsymbol{u}\in \mathbb{R}^4$:

$$\boldsymbol{u}=\left [ \begin{matrix}u_0\\u_1\\u_2\\u_3\end{matrix} \right ]$$

y averiguamos si es posible encontrar los escalares adecuados $k_1,k_2,k_3$ tales que
$$k_1\begin{bmatrix}1\\2\\3\\4\end{bmatrix} + k_2 \begin{bmatrix}2\\-1\\0\\2\end{bmatrix} + k_3\begin{bmatrix}1\\1\\0\\-4\end{bmatrix}=\left [ \begin{matrix}u_0\\u_1\\u_2\\u_3\end{matrix} \right ]$$

Esto nos llevará forzosamente a un sistema de ecuaciones lineales cuya parte aumentada será $\boldsymbol{u}$.



Escribe el código necesario para decidir si el conjunto es generador o no lo es.

```python
# Escribe el código aquí:

# Creación de vectores como objeto de sympy
v1 = Matrix([1, 2, 3, 4])
v2 = Matrix([2, -1, 0, 2])
v3 = Matrix([1, 1, 0, -4])

# Creación de la matriz de coeficientes
Ac = Matrix(np.hstack((v1,v2,v3)))

print('Matriz original:')
print(Ac)

# Creacion de los elementos del vector arbitrario con la variable u y subíndices
x,y,z,w = symbols('u:4')

# Creación de un vector arbitrario
va = Matrix([x,y,z,w])

# Creación de la matriz aumentada
Au = Matrix(np.hstack((v1,v2,v3,va)))

print('Matriz aumentada')
print(Au)

print('Matriz escalonada reducida:')
print(Ac.rref())

# Asignamos a una variable la dimensión que generan los vectores
Dimension = Ac.rank()

print('Rango de la matriz (Dimensión que generan):')
print(Dimension)

# Condición que evalúa si el rango alcanza la dimensión 4 que exige el espacio R^4
if Dimension == 4:
  print('Es un conjunto generador de R^4')
else:
  print('No es un conjunto generador de R^4')
```

### Ejercicio 3
Recordar que la dimensión del espacio vectorial $\mathbb{R}^n$  es precisamente $n$, por lo que cualquier subespacio propio $U\subsetneq\mathbb{R}^n$ deberá tener dimensión menor.

¿Cuál es la dimensión del subespacio $U$ generado por los siguientes vectores?

$$ \left \{
\boldsymbol{x}_1=\begin{bmatrix}1\\2\\-1\\-1\\-1\end{bmatrix}, \  \boldsymbol{x}_2=\begin{bmatrix}2\\-1\\1\\2\\-2\end{bmatrix}, \  \boldsymbol{x}_3=\begin{bmatrix}3\\-4\\3\\5\\-3\end{bmatrix}, \  \boldsymbol{x}_4=\begin{bmatrix}-1\\8\\-5\\-6\\1\end{bmatrix} \right\} \subset\mathbb{R}^5$$


*Solución*. Dado que los vectores dados son un conjunto generador, es necesario averiguar si son linealmente independientes; esto es, que la solución de la ecuación vectorial
$$\sum_{i=1}^4k_i\textbf{x}_i=\textbf{0}.$$
que conlleva al sistema de ecuaciones lineal homogeneo con matriz de coeficientes:

$$\begin{bmatrix}
1&2&3&-1\\
2&-1&-4&8\\
-1&1&3&-5\\
-1&2&5&-6\\
-1&-2&-3&1
\end{bmatrix}$$

```python
# Escribe tu código aquí:

# Creación de los vectores como objeto de sympy
v1 = Matrix([1,2,-1,-1,-1])
v2 = Matrix([2,-1,1,2,-2])
v3 = Matrix([3,-4,3,5,-3])
v4 = Matrix([-1,8,-5,-6,1])

# Creación de la matriz de coeficientes
A = Matrix(np.hstack((v1,v2,v3,v4)))

print(A)

# Creacion del vector con solución homogénea
solh = Matrix([0,0,0,0,0])

# Creación de la matriz aumentada
A = Matrix(np.hstack((v1,v2,v3,v4,solh)))

print(A)
print(A.rref())

# Cálculo de la dimension del espacio
Dimension = A.rank()

print('Dimension del espacio vectorial:')
print(Dimension)
```
